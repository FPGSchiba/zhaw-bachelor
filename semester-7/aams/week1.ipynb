```python
def linear(x, d):
    return x + d
```

```python
x_0 = 20
d = -3
num_iter = 10
x = x_0
for i in range(num_iter):
    x = linear(x, d)
    print(f"x_{i} = {x}")
```

```python
def logistic_map(x, r):
    return r * x * (1 - x)
```

```python
x_0 = 0.1
r = 4
num_iter = 9
x = x_0
for i in range(num_iter):
    print(f"x_{i} = {x}")
    x = logistic_map(x, r)
print(f"x_{num_iter} = {x}")
```
